```python
from ultralytics import YOLOE

# Initialize a YOLOE model
model = YOLOE("yoloe-11l-seg.pt")  # or select yoloe-11s/m-seg.pt for different sizes
```

```python
object_names = [
    "barcode",
    "shipping label",
    "QR code",
    "sign",
    "room sign",
    "handwritten text",
    "printed text",
    "whiteboard",
    "calendar",
    "person",
    "cardboard box",
    "painting",
    "poster",
    "portrait",
    "drawing",
    "photograph",
]
class_mapping = {v: k for k, v in zip(object_names, range(len(object_names)))}
print(class_mapping)
with open("yoloe_custom_mapping.txt", "w") as f:
    for k, v in class_mapping.items():
        f.write(f"{k}: {v}\n")

model.set_classes(object_names, model.get_text_pe(object_names))

model.save("yoloe-custom.pt")

# Run detection on the given image
# results = model.predict("../vggt/scene_images/9954589741/")
# results = model.predict("./test_images/")
```

```python
model = YOLOE("yoloe-barcode.pt")

results = model.predict("./test_images/")
```

```python

# Save results
count = 0
for result in results:
    count += 1
    result.save(filename="./yoloe_results/" + str(count) + ".jpg")
    print(result.probs)
    print(result.names)
```
